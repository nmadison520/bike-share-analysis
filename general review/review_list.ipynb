# What is a Python List?
> A list is a collection of items in a particular order. Lists can hold any data type and are mutable, meaning they can be changed.

```python

hokie_buildings = ["Burruss Hall", "Goodwin Hall", "Lane Stadium", "Torgersen Hall"]
```

#### Element Access, Modifying, and Looping

```python
print(hokie_buildings[0])
print(hokie_buildings[-1])
hokie_buildings.append('New Classroom Building')
print(hokie_buildings)
print(*hokie_buildings) # notice the * 
for building in hokie_buildings:
    building = "Squires"
print(hokie_buildings)
```

#### Useful List Methods
#### Useful List Methods

1. `.append()`  
   Adds an item to the end of the list.  
   Example: `hokie_buildings.append("Newman Library")`

2. `.extend()`  
   Adds individual items from another iterable (like another list).  
   Example: `hokie_buildings.extend(["Squires", "War Memorial"])`

3. `.sort()`  
   Sorts the list in ascending order (alphabetically or numerically).  
   Example: `hokie_buildings.sort()`

4. `.reverse()`  
   Reverses the order of the list.  
   Example: `hokie_buildings.reverse()`

5. `.remove(item)`  
   Removes the first occurrence of the specified item.  
   Example: `hokie_buildings.remove("Squires")`
