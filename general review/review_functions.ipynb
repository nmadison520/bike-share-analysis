# What Is a Function in Python?
> A function in Python is: 
>  - named
>  - self-contained block of code
>  - can accept input values (called parameters)
>  - execute a sequence of statements
>  - optionally return a result to the caller

> A function is a reusable block of code that performs a specific task. It helps<br>
> organize your program, avoid repetition, and make your code easier to read and maintain.

### Weather Advice
> The current weather_advice(temp, condition) function gives advice<br>
> based on temperature and weather condition, but it requires the<br>
> programmer to manually pass values.

Your task: Modify the function so that it asks the user for input using the input() function.



```python
def weather_advice(temp, condition):
    """Prints advice based on temperature and weather condition."""
    
    if temp < 0:
        print("It's freezing! Wear heavy layers.")
    elif temp < 10:
        print("Very cold. A coat and gloves are recommended.")
    elif temp < 20:
        print("Cool weather. A light jacket should be fine.")
    elif temp < 30:
        print("Warm. Dress comfortably.")
    else:
        print("Hot! Stay hydrated and wear sunscreen.")

    if condition == "rain":
        print("Don't forget your umbrella.")
    elif condition == "snow":
        print("Drive carefully and wear boots.")
    elif condition == "sunny":
        print("Enjoy the sunshine!")
    elif condition == "windy":
        print("Secure loose items and wear a windbreaker.")
    else:
        print("Check the forecast for more details.")
```
