# What is a Python Dictionary?
> A dictionary is a built-in Python data structure that stores data in key-value pairs.
>  - After Python 3.7, It's ordered, mutable, and indexed by keys
>  - Keys must be unique and immutable (like strings, numbers, or tuples).

```python
# here is an python dictionary example
hokie_student = {
    "name": "Gobble T. Hokie",
    "student_id": "VT007",
    "major": "Chicken Engineering",
    "age": 19,
    "year": "Sophomore",
    "favorite_food": "West End Mac & Cheese",
    "hobbies": ["Tailgating", "Jumping to 'Enter Sandman'", "Chasing squirrels", "Inventing turkey calls"],
    "GPA": 2.718,  # Because math is hard but Euler is cool
    "dorm": "Pritchard Hall",
    "mode_of_transport": "Skateboard with turkey feathers",
    "quirks": {
        "never_misses": "Home football games",
        "always_carries": "A lunchbox full of maroon and orange jellybeans",
        "secret_talent": "Can gobble in 7 languages"
    },
    "is_hokie_for_life": True
}
```
