# Creating an Animation (.gif)

```python
import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image
import os

# Create a directory to store the images
os.makedirs("temp_images", exist_ok=True)
```

### Load the Dataset

```python

weather = pd.read_csv("weather.csv")
bburg_weather = weather[weather['city'] == 'Blacksburg'].copy()
anch_weather = weather[weather['city'] == 'Anchorage'].copy()
```

### Data Analysis
>- Find average temperature for each day
>    - We were not given an average temperature with the dataset.
>    - We will average the min, max, morning, afternoon, evening, and night temps to give an estimated average temp.
>- Create a map to change the month from a number to an abbrev month name
>- Group the data by month and aggregate mean

```python
bburg_weather['temp avg'] = bburg_weather[["temp min", "temp max", "temp morning", "temp afternoon", "temp evening", "temp night"]].mean(axis=1)
yearly_temps = bburg_weather.groupby(['year', 'month']).agg({"temp avg" : "mean"}).reset_index()
months = {1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr", 5: "May", 6: "Jun", 7: "Jul", 8: "Aug", 9: "Sep", 10: "Oct", 11: "Nov", 12:"Dec"}
yearly_temps["month name"] = yearly_temps['month'].map(months)
# Here is where you might filter the data if you wanted to represent a 
#  "restricted" average temp
avg_temps = yearly_temps.groupby('month')['temp avg'].mean().reset_index()
avg_temps['month name'] = avg_temps['month'].map(months)
```

### Creating Lots of Images

```python
#print(plt.style.available)
plt.style.use('seaborn-v0_8-darkgrid')

image_files = []
for i in range(1980, 2025):
    fig, ax = plt.subplots(figsize=(8,4))
    y = yearly_temps[yearly_temps['year']== i]
    ax.plot(y['month name'], avg_temps['temp avg'], marker='.', label='Average')
    ax.plot(y['month name'], y['temp avg'], marker='.', label=i)
    ax.set_ylim([0, 90])
    
    ax.set_title(f"Blacksburg Average Temperatures (F)\nAverage VS {i}")
    ax.set_ylabel("Average Temperature (F)")
    ax.legend(loc='upper right', bbox_to_anchor=(1.1, 1))
    
    file_name = f"temp_images/line{i - 1980}.png"
    plt.savefig(file_name, bbox_inches='tight')
    plt.close()
    
    image_files.append(file_name)
```

### Creating the Animation (.gif)

```python
frames = [Image.open(img) for img in image_files]
gif_filename = "temperature_animation.gif"
frames[0].save(
    gif_filename,
    format='GIF',
    append_images=frames[1:], #start at 1 since 0 is already used
    save_all=True,
    duration=600,
    loop=0
)

print(f"Animated GIF saved as {gif_filename}")
```
