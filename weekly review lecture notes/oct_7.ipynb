### Abstract Classes
>- An abstract class in Python is a blueprint for other classes. It defines a common interface and structure that its subclasses must follow.
>- Think of it as a template that outlines what a class should do, but leaves the details to be filled in by subclasses.

**Why use abstract classes?**
>- Abstract classes are used when you want to ensure that certain methods are implemented in all derived classes, without providing a complete implementation in the base class.

**Key Characteristics of Abstract Classes**
>- **Cannot** Be Instantiated
>- Defined Using the abc Module
>- Conatins Abstract Methods
>- Used to Enforce a Contract
>- Supports Polymorphism

### Using the abc module and ABC base class:

```python
# Example of:
#  defined using the abc module
#  contains abstract methods
#  used to enforce a contract
#  supports polymorphism

from abc import ABC, abstractmethod

class Student(ABC):
    def __init__(self, name: str):
        self.name = name
    
    def __str__(self):
        return f"{self.name}".lower()

    @abstractmethod
    def submit_assignment(self, assignment: str):
        pass
```

```python
# Defining subclasses of the abstract class Student
class OnlineStudent(Student):
    def submit_assignment(self, assignment: str): 
        print(f"{self.name} uploads {assignment} to Canvas.")

class InPersonStudent(Student):
    def submit_assignment(self, assignment: str):
        print(f"{self.name} hands in a printed copy of {assignment}.")

class ErrorStudent(Student):
    # This class is missing the required abstract method 'submit_assignment'
    # and will raise an error if instantiated.
    def greet(self, msg: str):
        print(f"Hello! I'm {self.name}.\n{msg}")

student1 = OnlineStudent("Sam")
student2 = InPersonStudent("Sung")
# uncomment to see the error!
#student3 = ErrorStudent("Sad") # produces a TypeError

print("Student 1:", end=' ')
student1.submit_assignment("Project 1")
print("Student 2:", end=' ')
student2.submit_assignment("Classwork 5")
```

### Practice Problem

>- You are building a wildlife tracking system for a nature reserve. The system needs to track different types of animals and their movements. To ensure consistency, you decide to use an abstract base class.
Instructions
>
>
>- Create an abstract class called Animal with the following:

>    - An `__init__` method that takes name and species.
>    - An `__str__` method that returns a string with the animal's name and species.
>    - An abstract method track_movement() that must be implemented by subclasses.
>
>- Create Two Subclasses:
>    - Bird: Implements track_movement() by returning a string like "Bird {name} is flying at altitude {altitude} meters."
>    - Squirrel: Implements track_movement() by returning a string like "Squirrel {name} is walking {distance} meters."
>
>- Demonstrate Usage:
>    - Create instances of Bird and Mammal.
>    - Call get_info() and track_movement() on each instance.

```python
# Practice Problem
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name, species):
        self.name= name
        self.species= species
    def __str__(self):
        return f"{self.name} ({self.species})"

    @abstractmethod
    def track_movement(self):
        pass
class Bird(Animal):
    def __init__(self, name, altitude):
        super().__init__(name, "Bird")
        self.altitude=altitude

    def track_movement(self):
        return f"Bird {self.name} is flying at altitude {self.altitude} meters"

class Squirrel(Animal):
    def __init__(self, name, distance):
        super().__init__(name, "Squirrel")
        self.distance=distance

    def track_movement(self):
        return f"Squirrel {self.name} is walking {self.distance} meters"
```
