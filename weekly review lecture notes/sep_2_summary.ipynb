### Exploring For and While Loops
1. Can we change a for loop that does exactly 3 iterations into a while loop?
2. When should we use a for loop, when should we use a while loop?
3. Here is an example of when a while loop is appropriate instead of using a for loop.

```python
for i in range(3):
    print(f'{i}: Turkey')

count = 0
while (count < 3):
    print(f'{count}: Turkey')
    count += 1

guess = 0
while (guess != 1):
    guess = int(input("Guess a number: "))
```
